# Análise de Dados de Vendas

Projeto de análise exploratória de dados utilizando Python e pandas.


```python
import os

# Criar a pasta dados dentro de /content
os.makedirs('/content/dados', exist_ok=True)

# Criar o arquivo vendas.csv
with open('/content/dados/vendas.csv', 'w') as f:
    f.write(
        "data,produto,categoria,quantidade,preco,regiao\n"
        "2024-01-05,Notebook,Eletrônicos,2,3500,Sudeste\n"
        "2024-01-06,Mouse,Eletrônicos,10,80,Sudeste\n"
        "2024-01-07,Cadeira,Moveis,1,900,Sul\n"
        "2024-01-08,Teclado,Eletrônicos,5,150,Sudeste\n"
        "2024-01-09,Mesa,Moveis,2,1200,Nordeste\n"
        "2024-01-10,Monitor,Eletrônicos,3,1100,Sul\n"
        "2024-01-11,Notebook,Eletrônicos,1,3500,Nordeste\n"
        "2024-01-12,Mouse,Eletrônicos,7,80,Sudeste\n"
    )

print("Arquivos em /content:", os.listdir('/content'))
print("Arquivos em /content/dados:", os.listdir('/content/dados'))
```

```python
import os

print("Pasta atual:", os.getcwd())
print("Conteúdo da pasta atual:", os.listdir())
```

```python
import pandas as pd

df = pd.read_csv('/content/dados/vendas.csv')
df.head()
```

```python
df['faturamento'] = df['quantidade'] * df['preco']
df
```

```python
df['faturamento'].sum()

```

```python
faturamento_categoria = df.groupby('categoria')['faturamento'].sum()
faturamento_categoria
```

```python
import matplotlib.pyplot as plt

faturamento_categoria.plot(kind='bar')
plt.title('Faturamento por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Faturamento')
plt.show()
```

## Conclusão

A análise dos dados de vendas mostrou que a categoria **Eletrônicos** vendeu mais,
sendo o principal responsável pela receita da loja no período analisado.
Já a categoria **Móveis** teve participação menor, indicando uma possível oportunidade de crescimento.
